# 01. Exploratory Data Analysis: NASA POWER & Global Wind Atlas

## Overview
This notebook explores surface solar irradiance metrics from **NASA POWER** Agro-Climatology datasets, 100m hub-height wind speed estimates from the **Global Wind Atlas (GWA)**, and topographic rasters from **SRTM** across candidate renewable energy deployment sites in India.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting aesthetics
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 6)
print("Libraries successfully imported!")
```

## 1. Load Meteorological Data
Simulating candidate grid coordinates across Northern, Western, Southern, and Central Indian energy corridors.

```python
np.random.seed(42)
n_samples = 1000

lats = np.random.uniform(8.0, 32.0, n_samples)
lngs = np.random.uniform(68.0, 88.0, n_samples)

# Solar GHI (kWh/m2/day)
solar_ghi = np.round(np.random.normal(5.4, 0.8, n_samples).clip(3.0, 7.5), 2)

# Wind speed at 100m (m/s)
wind_speed_100m = np.round(np.random.normal(6.2, 1.5, n_samples).clip(2.5, 11.0), 2)

# Terrain slope (degrees)
slope_deg = np.round(np.random.exponential(2.5, n_samples).clip(0.5, 18.0), 1)

# Elevation (meters)
elevation_m = np.round(np.random.uniform(50, 800, n_samples), 0)

df = pd.DataFrame({
    'latitude': lats,
    'longitude': lngs,
    'solar_ghi_kwh_m2_day': solar_ghi,
    'wind_speed_100m_ms': wind_speed_100m,
    'slope_deg': slope_deg,
    'elevation_m': elevation_m
})

df.head()
```

## 2. Statistical Distribution Analysis
Inspecting distributions of key environmental parameters.

```python
df.describe()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

sns.histplot(df['solar_ghi_kwh_m2_day'], kde=True, ax=axes[0], color='gold')
axes[0].set_title('Solar Irradiance (GHI) Distribution')
axes[0].set_xlabel('kWh/m²/day')

sns.histplot(df['wind_speed_100m_ms'], kde=True, ax=axes[1], color='teal')
axes[1].set_title('Wind Speed (100m Hub) Distribution')
axes[1].set_xlabel('m/s')

sns.histplot(df['slope_deg'], kde=True, ax=axes[2], color='coral')
axes[2].set_title('Terrain Slope Distribution')
axes[2].set_xlabel('Degrees')

plt.tight_layout()
plt.show()
```

## 3. Correlation Heatmap
Analyzing pairwise relationships between geographical and climate metrics.

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(), annot=True, cmap='Blues', fmt='.2f')
plt.title('Feature Correlation Matrix')
plt.show()
```

## 4. Summary Findings
- High GHI (> 5.5 kWh/m²/day) is concentrated in Western and Northern India (Rajasthan, Gujarat).
- Strong wind velocities (> 7.0 m/s) correlate with coastal and elevated interior ridges.
- Terrain slope exhibits exponential distribution; filtering slope > 15° removes non-viable cliff parcels.
